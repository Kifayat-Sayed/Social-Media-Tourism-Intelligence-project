# ✈️ Social Media Tourism – Customer Purchase Propensity Prediction

## 📌 1. Problem Statement

An aviation company offering domestic and international flight services wants to shift from traditional telecalling-based customer acquisition to a more targeted, data-driven digital marketing approach.

To achieve this, the company has partnered with a social networking platform to understand customers' digital behaviour, travel interests, and purchasing patterns.

The objective is to leverage customer data and machine learning to identify individuals with a **high likelihood of purchasing flight tickets** and target them with personalised advertisements, exclusive discounts, and loyalty rewards.

## 🎯 2. Need for the Study / Project

As businesses increasingly move towards digital marketing, there is a growing need to use data analytics and machine learning to generate high-quality leads and improve customer acquisition.

Since digital advertisements involve significant costs, it is essential to identify the right audience and ensure that marketing campaigns reach customers who are more likely to convert.

Customer purchase propensity may depend on several factors, including:

- 📱 **Login Devices:** Devices and platforms customers use to access social media and travel websites.
- 👨‍👩‍👧‍👦 **Family Information:** Number of family members and group travel preferences.
- 📍 **Location Preferences:** Preferred destinations, travel locations, and types of places customers are interested in visiting.
- 🌍 **Historical Website Visits:** Previous visits, browsing frequency, and interactions with travel-related websites.
- 🛒 **Digital Behaviour:** Customer engagement, interests, online activities, and interactions with advertisements.
- ✈️ **Travel Preferences:** Previous travel history, destination preferences, and booking patterns.

By analysing these factors, the company can understand customer behaviour, predict purchase likelihood, and deliver personalised offers such as discounts, promotional deals, and loyalty bonuses at the right time.

## 💼 3. Understanding Business & Social Opportunities

### 🚀 Business Opportunities

- **Targeted Customer Acquisition:** Identify potential customers who have a higher likelihood of purchasing flight tickets through data-driven targeting.
- **Cost-Effective Marketing:** Reduce unnecessary advertising expenditure by focusing campaigns on high-propensity customers.
- **Faster Lead Generation:** Generate quality leads in less time compared to traditional telecalling methods.
- **Personalised Marketing:** Offer customised discounts, promotional campaigns, and travel packages based on individual customer preferences.
- **Customer Retention:** Build long-term relationships through loyalty programmes, exclusive benefits, and personalised travel experiences.
- **Revenue Growth:** Improve advertisement conversion rates, increase ticket bookings, and maximise overall business revenue.

### 🤝 Social Opportunities

- **Strengthening Customer Relationships:** Build stronger connections with existing customers through personalised communication and exclusive benefits.
- **Referral & Reward Programmes:** Encourage existing customers to recommend the airline to friends and family through referral bonuses and discounts.
- **Improved Customer Engagement:** Provide relevant travel offers, destination recommendations, and promotional content based on customers' interests.
- **Customer Loyalty:** Encourage repeat bookings by rewarding loyal customers with special discounts, loyalty points, and travel benefits.
- **Better Customer Experience:** Make travel offers more relevant and accessible by understanding individual preferences and travel behaviour.

```python
fig = plt.figure(figsize=(15,15))
fig.subplots_adjust(hspace=0.5, top=0.9, wspace= 0.1, bottom = 0.1)
fig.suptitle('Outlier Analysis using BoxPlots', fontsize=20)
fig.tight_layout()

for i, col in enumerate(cleaned_df.select_dtypes(include=['float64','int64']).columns):
    ax1 = fig.add_subplot(5,3,i+1)
    sns.boxplot(x = cleaned_df[col], ax = ax1)
    ax1.set_xlabel(col)
    ax1.title.set_text(f'{col} \nSkew :- {round(cleaned_df[col].skew(),2)}')

plt.suptitle('Boxplot before outlier treatment', fontsize = 25)
plt.show();
```

```python
def get_outlier_range(cname):
    sorted(cname)
    q1, q3 = cname.quantile([0.25,0.75])
    IQR = q3-q1
    llimit = q1-(1.5 * IQR)
    hlimit = q3+(1.5 * IQR)
    return llimit, hlimit
```

```python
clist = ['Yearly_avg_view_on_travel_page','total_likes_on_outstation_checkin_given','Yearly_avg_comment_on_travel_page',
         'total_likes_on_outofstation_checkin_received', 'montly_avg_comment_on_company_page',
         'Daily_Avg_mins_spend_on_traveling_page']
for col in clist:
    lval, hval = get_outlier_range(df[col])
    cleaned_df[col] = np.where(cleaned_df[col] >hval, hval, cleaned_df[col])
    cleaned_df[col] = np.where(cleaned_df[col] <lval, lval, cleaned_df[col])
```

```python
correlation_threshold = 0.5

def high_cor_function(df):
    cor = df.corr()
    c1 = cor.stack().sort_values(ascending=False).drop_duplicates()
    high_cor = c1[c1.values!=1]
    return (high_cor[high_cor>correlation_threshold])
```

#### Making Elbow plot (up to n=10) and identify optimum number of clusters for k-means algorithm.

```python
from sklearn.cluster import KMeans
```

```python
#Empty dictionary to store the SSE (sum of the squared Euclidean distances) for each value of k
sse = {}

# iterate for a range of Ks and fit the scaled data to the algorithm. Use inertia attribute from the clustering object and
# store the inertia value for that k
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=123).fit(scaled_df)
    sse[k] = kmeans.inertia_

#Elbow plot
plt.figure()
sns.pointplot(x=list(sse.keys()), y=list(sse.values()))
plt.xlabel("Number of cluster")
plt.ylabel("SSE")
plt.show();
```

#### Silhouette scores for up to 10 clusters and identify optimum number of clusters.

```python
from sklearn.metrics import silhouette_samples, silhouette_score
```

```python
#Empty dictionary to store the Silhouette score for each value of k
sc = {}

# iterate for a range of Ks and fit the scaled data to the algorithm. Store the Silhouette score for that k
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=123).fit(scaled_df)
    labels = kmeans.predict(scaled_df)
    sc[k] = silhouette_score(scaled_df, labels)

#Elbow plot
plt.figure()
sns.pointplot(x=list(sc.keys()), y=list(sc.values()))
plt.xlabel("Number of cluster")
plt.ylabel("Silhouette Score")
plt.show();
```

**a) Is the data unbalanced? If so, what can be done? Please explain in the context of the business**
   - Yes, we can see that we have 16% of users who have purchased the product whereas 84% of users did not purchase the product. So, we can confirm that data is unbalanced
   - We can use Synthetic Minority Oversampling Technique (SMOTE) to create syntehtic data from minority samples. This will help us create a training set which is nearly class-balanced. And then this training set can be used to train on Classifier Models. We need to keep in mind that we do not use Synthetic data for test or validation test. Test and Validation set always consists of our original data to understand or evaluate our models performance.
   - Usually the predictor assigns the “1” for values with probability greater than or equal to 0.5 and “0” for probability values less than 0.5. This is the default threshold that is used by models. But the power of the model is lesser due to imbalanced data for the target variables. Hence, we can choose the cutoff which will give more descriptive power to the models.
       - We can use the ROC Curve to find the optimal threshold. For this, we will fetch TPR, FPR and Threshold from the ROC Curve. Then we will get the argmax of TPR and FPR to find the optimal threshold values.
       - Increasing the cut-off leads to more False Negatives and decreasing the cut-off leads to more False Positives.
       - Hence, this threshold cut-off values must be chose wisely.
       
   - From business point of view, it is important to know the customers possibility of conversion.
       - For example if the probability of customer purchasing a product is 20%. At this point, we can't be sure whether a customer will buy the product or not.
       - Let us take another example, if the probability of customer purchasing a product is 35%. At this point, we can't be sure whether a customer will buy the product or not but business would be keen to invest some time and money through one to one digital marketing or offers to convert the customers mind to purchase the product.
       - From business point of view, we need to find out probable leads that can be valueable to business in increasing their revenue and sales.
